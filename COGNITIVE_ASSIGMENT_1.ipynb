```python
def addprime(n):
  total=0
  for i in range(2,n + 1):
    prime=True
    for j in range(2,i):
      if i%j==0:
        prime=False
        break
    if prime:
      total=total+i

  return total
print(addprime(10))
print(addprime(15))
```
