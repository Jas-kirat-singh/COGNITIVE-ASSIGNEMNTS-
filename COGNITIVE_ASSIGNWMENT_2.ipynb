```python
#Q3)
import random as rd
rd.seed(1024170312)
numbers=[rd.randint(100,900) for i in range(100)]
print(numbers)

```

```python
odd=[]
for x in numbers:
  if x%2!=0:
    odd.append(x)
print("odd numbers:")
print(odd)
print("number of odd:",len(odd))
```

```python
even=[]
for x in numbers:
  if x%2==0:
    even.append(x)
print("even numbers:")
print(even)
print("number of even:",len(even))
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True



prime_numbers = [num for num in numbers if is_prime(num)]


print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))
```

```python
maxcount=max(numbers,key=numbers.count)
print("Most frequent:",maxcount)
print("Number of times:",numbers.count(maxcount))
```

```python
roll=[1,0,2,4,1,7,0,3,1,2]
print(roll)
A={int(x)*7 for x in roll[:8]}
print(A)
B={int(x)*9 for x in roll[:8]}
print(B)
```

```python
union_set=A.union(B)
print(union_set)
```

```python
intersect=A.intersection(B)
print(intersect)
```

```python
a_minus_b=A.difference(B)
print("A minus B:",a_minus_b)
b_minus_a=B.difference(A)
print("B minus A:",b_minus_a)
a_sym_b=A.symmetric_difference(B)
print("A sym B:",a_sym_b)
```

```python
print("A is subset of B:",A.issubset(B))
print("B is supperset of A:",B.issuperset(A))
```

```python
print(A)
val=int(input("Enter val:"))
if val in A:
  A.discard(val)

print(A)
#discard() is safer because it does not give key error when element is not presnet
#in the set unlike remove()
```

```python
my_dict = { "name": "Jaskirat",
           "roll_no": "1024170312",
            "branch": "COPC",
            "age": 20,
            "city": "Patiala" }
```

```python
my_dict["location"]=my_dict.pop("city")
print(my_dict)
```

```python
my_dict["CGPA"]=9.52
my_dict["age"]=my_dict["age"]+1
print(my_dict)
```

```python
copy1=my_dict.copy()
copy2=my_dict.copy()
#pop
val=copy1.pop("branch")
print(val)
print(copy1)
del copy2["branch"]
print(copy2)
#pop() removes the key and returns the values of the key whereas del deletes key and value and not return anything
```

```python
for key,value in my_dict.items():
  print(key,"->",value)
```
