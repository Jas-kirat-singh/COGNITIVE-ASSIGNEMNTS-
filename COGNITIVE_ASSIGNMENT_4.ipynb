```python
fixed_entries = [
{"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
"keywords": "fee cost price charge", "category": "billing"},
{"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
"keywords": "password reset login", "category": "account"},
{"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
"keywords": "hours timing open time", "category": "general"},
{"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
"keywords": "pay payment upi fee", "category": "billing"},
]
```

```python
categories = ["billing", "account", "general"]

last_two = [int(d) for d in str(roll)[-2:]]

custom_entries = []

for d in last_two:

    category = categories[d % 3]

    if category == "billing":

        question = "how can i check my payment status"
        answer = "You can check your payment status in the billing section."
        keywords = "payment status billing"

    elif category == "account":

        question = "how do i update my registered mobile number"
        answer = "Go to Account Settings and update your registered mobile number."
        keywords = "mobile number account update"

    else:

        question = "where can i find the latest service updates"
        answer = "You can find the latest service updates on the official updates page."
        keywords = "updates service latest"

    custom_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })
```

```python
final_entries=fixed_entries+custom_entries
df=pd.DataFrame(final_entries)
df
```

```python
def score_query(query, df):


    query_words = query.lower().split()

    results = []

    for index, row in df.iterrows():

        keywords = row["keywords"].lower().split()


        score = sum(1 for word in keywords if word in query_words)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })


    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return results

query = input("Enter your query: ")

results = score_query(query, df)

print(results)
```

```python
def score_query_with_ties(query, df):

    query_words = query.lower().split()

    results = []


    for index, row in df.iterrows():

        keywords = row["keywords"].lower().split()

        score = sum(1 for word in keywords if word in query_words)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })


    if not results:
        print("No matching FAQ found.")
        return

    highest_score = max(result["score"] for result in results)


    best_matches = [
        result for result in results
        if result["score"] == highest_score
    ]


    if len(best_matches) > 1:

        print("\nTie detected!")
        print("Highest score:", highest_score)

        for result in best_matches:
            print("\nQuestion:", result["question"])
            print("Answer:", result["answer"])
            print("Category:", result["category"])

    else:

        print("\nBest match:")
        print("Question:", best_matches[0]["question"])
        print("Answer:", best_matches[0]["answer"])
        print("Category:", best_matches[0]["category"])
```

```python
query1 = input("Enter a query that may produce a tie: ")

score_query_with_ties(query1, df)

```
